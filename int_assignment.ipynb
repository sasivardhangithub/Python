```python
print("Question 1: Calculate the product of first 10 natural numbers")
product = 1
for num in range(1, 11):
    product *= num
print("Product of first 10 natural numbers:", product)
```

```python
n = 12345
digit = 0
while n > 0:
    digit += n % 10 
    n //= 10
print("The sum of the digits of 12345 is", digit)
```

```python
n = 97

if n>1:
    for i in range(2, int(n/2)+1):
        if (n % i) == 0:
            print(n, "is not a prime number")
            break
    else:
        print(n, "is a prime number")
```

```python
n = 8
fact = 1
while n > 1:
    fact *= n
    n -= 1
print("\nQuestion 7: Calculate the factorial of 8")
print("The factorial of 8 is", fact)
```

```python
numbers_lst = [15, 23, 31, 42, 56]
average = sum(numbers_lst) / len(numbers_lst)
print("\nQuestion 8: Calculate the average of the numbers in the list [15, 23, 31, 42, 56]")
print("The average of the numbers in the list is", average)
```
