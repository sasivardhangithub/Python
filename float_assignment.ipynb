```python
print("Question 1: Calculate the average of 3.14, 2.718, 1.618, 0.577")
list_of_numbers = [3.14, 2.718, 1.618, 0.577]
average = sum(list_of_numbers) / len(list_of_numbers)
print("The average of the numbers is", average)
```

```python
print("\nQuestion 2: Convert 98.6 Fahrenheit to Celsius")
farhenheit = 98.6
celisus = (farhenheit - 32) * 5 / 9
print("The Celsius equivalent of 98.6 Fahrenheit is", celisus)
```

```python
print("\nQuestion 3: Calculate compound interest on $1000 at 5.5% for 3 years")
principal = 1000
rate  = 5.5 / 100
time = 3
compound_interest = principal * (1 + rate) ** time - principal
print("The compound interest on $1000 at 5.5% for 3 years is", compound_interest)
```

```python
print("\nQuestion 4: Find the hypotenuse of a right triangle with sides 3.5 and 4.2")
side1 = 3.5
side2 = 4.2
hypotenuse = (side1 ** 2 + side2 ** 2) ** 0.5
print("The hypotenuse of the right triangle is", hypotenuse)
```

```python
print("\nQuestion 5: Calculate the volume of a sphere with radius 7.8")
radius = 7.8
volume = (4/3) * 3.14*(radius ** 3)
print("The volume of the sphere with radius 7.8 is", volume)
```

```python
print("\nQuestion 6: Round 3.14159 to 3 decimal places")
number = 3.14159
rounded_number = round(number, 3)
print("The number 3.14159 rounded to 3 decimal places is", rounded_number)
```

```python
print("\nQuestion 7: Calculate the percentage: 45 out of 67")
percentage = (45 / 67) * 100
print("The percentage of 45 out of 67 is", percentage)
```

```python
print("\nQuestion 8: Find the square root of 23.456")
import math
number = 23.456
sqrt = math.sqrt(number)
print("The square root of 23.456 is", sqrt)
```

```python
print("\nQuestion 9: Calculate simple interest: Principal=2500, Rate=6.5%, Time=2.5 years")
principal = 2500
rate = 6.5 /100
time = 2.5
simple_interest = principal * rate * time
print("Simple interest is", simple_interest)
```
