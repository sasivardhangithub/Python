Write a program to find the Greatest Common Divisor (GCD) of two numbers using the Euclidean algorithm.

Algorithm:
1. Divide the larger number by the smaller number
2. Replace the larger number with the smaller number
3. Replace the smaller number with the remainder
4. Repeat until the remainder is 0
5. The GCD is the last non-zero remainder

**Hint:** While b != 0: store remainder (a%b), then a=b, b=remainder. When loop ends, a is the GCD.

````python
# Your code here
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

while b != 0:
    remainder = a % b
    a = b
    b = remainder

print("GCD:", a)
````

### Problem 6.6: Strong Number Checker (Hard)

Write a program to check if a number is a Strong number.

A Strong number is a number whose sum of factorial of digits is equal to the original number.

Example: 145 is a strong number because 1! + 4! + 5! = 1 + 24 + 120 = 145

**Hint:** Extract each digit, calculate its factorial, add to sum. Compare sum with original number.

````python
# Your code here
n = int(input("Enter a number: "))

original = n
total = 0

while n > 0:
    digit = n % 10

    factorial = 1

    for i in range(1, digit + 1):
        factorial *= i

    total += factorial
    n //= 10

if total == original:
    print("Strong number")
else:
    print("Not a strong number")
````

### Problem 7.1: Collatz Conjecture (Medium)

The Collatz conjecture is one of the most famous unsolved problems in mathematics. The rules are:
1. Start with any positive integer n
2. If n is even, divide it by 2
3. If n is odd, multiply it by 3 and add 1
4. Repeat until you reach 1

Write a program that:

- Counts how many steps it took

Example: Starting with 6: 6 → 3 → 10 → 5 → 16 → 8 → 4 → 2 → 1 (8 steps)

````python
# Your code here
n = int(input("Enter a positive integer: "))

steps = 0

print("Sequence:")

while n != 1:
    print(n, end=" → ")

    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1

    steps += 1

print(1)
print("Number of steps:", steps)
````

### Problem 7.2: Caesar Cipher (Hard)

Write a program that implements a Caesar cipher to encrypt and decrypt messages.

The Caesar cipher shifts each letter by a fixed number of positions in the alphabet.

Example: With shift=3
- 'A' becomes 'D'
- 'Hello' becomes 'Khoor'

Your program should:
1. Ask for a message and shift value
2. Provide options to encrypt or decrypt
3. Handle both uppercase and lowercase letters
4. Keep non-alphabetic characters unchanged

**Hint:** Loop through each character. Use ord() to get ASCII value, shift it, use chr() to convert back. Use modulo 26 for wrapping. Check if char.isalpha().

````python
# Your code here
message = input("Enter message: ")
shift = int(input("Enter shift value: "))

choice = input("Enter E for Encrypt or D for Decrypt: ").upper()

result = ""

for ch in message:

    if ch.isalpha():

        if ch.isupper():
            start = ord('A')
        else:
            start = ord('a')

        if choice == "E":
            new_char = chr((ord(ch) - start + shift) % 26 + start)

        elif choice == "D":
            new_char = chr((ord(ch) - start - shift) % 26 + start)

        else:
            print("Invalid choice")
            break

        result += new_char

    else:
        result += ch

print("Result:", result)
````

````python
# Your code here
balance = 10000
pin = 1234
history = []

attempts = 3

while attempts > 0:

    entered_pin = int(input("Enter your PIN: "))

    if entered_pin == pin:

        while True:

            print("\n--- ATM MENU ---")
            print("1. Check Balance")
            print("2. Withdraw Money")
            print("3. Deposit Money")
            print("4. Change PIN")
            print("5. Exit")

            choice = input("Enter your choice: ")

            if choice == "1":

                print("Balance:", balance)
                history.append(f"Balance checked: {balance}")

            elif choice == "2":

                amount = float(input("Enter withdrawal amount: "))

                if amount <= 0:
                    print("Invalid amount")

                elif amount > balance:
                    print("Insufficient balance")

                else:
                    balance -= amount
                    print("Withdrawal successful")
                    print("Remaining balance:", balance)

                    history.append(f"Withdrawn: {amount}")

            elif choice == "3":

                amount = float(input("Enter deposit amount: "))

                if amount <= 0:
                    print("Invalid amount")

                else:
                    balance += amount
                    print("Deposit successful")
                    print("Current balance:", balance)

                    history.append(f"Deposited: {amount}")

            elif choice == "4":

                new_pin = int(input("Enter new PIN: "))
                pin = new_pin

                print("PIN changed successfully")
                history.append("PIN changed")

            elif choice == "5":

                print("\nTransaction History:")

                for transaction in history[-5:]:
                    print(transaction)

                print("Thank you for using ATM")
                break

            else:
                print("Invalid choice")

        break

    else:

        attempts -= 1
        print("Incorrect PIN")
        print("Attempts remaining:", attempts)

        if attempts == 0:
            print("Card blocked")
````

````python
# Your code here
board = []

print("Enter 9 rows of Sudoku:")

for i in range(9):
    row = list(map(int, input().split()))
    board.append(row)


valid = True

# Check rows
for i in range(9):

    if set(board[i]) != set(range(1, 10)):
        valid = False
        break


# Check columns
if valid:

    for j in range(9):

        column = []

        for i in range(9):
            column.append(board[i][j])

        if set(column) != set(range(1, 10)):
            valid = False
            break


# Check 3x3 boxes
if valid:

    for row_start in range(0, 9, 3):

        for col_start in range(0, 9, 3):

            box = []

            for i in range(row_start, row_start + 3):

                for j in range(col_start, col_start + 3):
                    box.append(board[i][j])

            if set(box) != set(range(1, 10)):
                valid = False
                break

        if not valid:
            break


if valid:
    print("Valid Sudoku")
else:
    print("Invalid Sudoku")
````
